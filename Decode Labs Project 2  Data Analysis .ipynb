## Basic Statistics

Calculate the **count, mean, and median** for the numerical columns, as required for Project 2.

```python
# Select numerical columns and exclude OrderID because it is an identifier
numerical_cols = df.select_dtypes(include="number").columns.tolist()
numerical_cols = [col for col in numerical_cols if col != "OrderID"]

basic_stats = pd.DataFrame({
    "Count": df[numerical_cols].count(),
    "Mean": df[numerical_cols].mean(),
    "Median": df[numerical_cols].median()
}).round(2)

basic_stats
```

## Identify Trends and Distributions

Compare the mean and median to understand the general shape of the numerical variables.

```python
for col in numerical_cols:
    mean = df[col].mean()
    median = df[col].median()

    if mean > median:
        trend = "Right-skewed tendency"
    elif mean < median:
        trend = "Left-skewed tendency"
    else:
        trend = "Approximately symmetrical"

    print(f"{col}: Mean = {mean:.2f}, Median = {median:.2f} → {trend}")
```

## Identify Outliers

Use the **IQR method** to identify unusually high or low values.

```python
def find_outliers(data, column):
    q1 = data[column].quantile(0.25)
    q3 = data[column].quantile(0.75)
    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outliers = data[(data[column] < lower_limit) | (data[column] > upper_limit)]
    return outliers, lower_limit, upper_limit

outlier_summary = []

for col in numerical_cols:
    outliers, lower, upper = find_outliers(df, col)

    outlier_summary.append({
        "Column": col,
        "Lower Limit": round(lower, 2),
        "Upper Limit": round(upper, 2),
        "Outliers": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)
outlier_summary
```
